```python

# ============================================================
# ANÁLISE EXPLORATÓRIA DE DADOS — SALES TRANSACTIONS 2022–2025
# Baseado na apostila de Pré-processamento e Análise de Dados
# Execução: Google Colab
# ============================================================

# ETAPA 1 — IMPORTAR BIBLIOTECAS
# Bibliotecas são ferramentas prontas para cálculos e gráficos.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from google.colab import drive, files

pd.set_option("display.max_columns", 50)
pd.set_option("display.max_rows", 100)

print("Bibliotecas carregadas com sucesso!")


# ============================================================
# ETAPA 2 — CONECTAR O GOOGLE DRIVE
# ============================================================

drive.mount("/content/drive")


# ============================================================
# ETAPA 3 — INFORMAR O CAMINHO DO ARQUIVO
#
# ATENÇÃO: ALTERE SOMENTE A LINHA ABAIXO.
# Cole entre as aspas o caminho copiado do seu Drive.
# ============================================================

CAMINHO_ARQUIVO = "/content/drive/MyDrive/Banco de Dados 1/Sales_transactions_2022_2025.xlsx"
# ============================================================
# FIM DA PARTE QUE VOCÊ PRECISA ALTERAR
# ============================================================


# ETAPA 4 — CARREGAR A PLANILHA

df_original = pd.read_excel(CAMINHO_ARQUIVO)

# Criamos uma cópia para preservar os dados originais.
df = df_original.copy()

print("\nARQUIVO CARREGADO COM SUCESSO!")
print("Quantidade de linhas:", df.shape[0])
print("Quantidade de variáveis:", df.shape[1])

print("\nPrimeiras cinco observações:")
display(df.head())


# ============================================================
# ETAPA 5 — IDENTIFICAR E CLASSIFICAR AS VARIÁVEIS
# ============================================================

# A classificação abaixo leva em conta o significado
# das variáveis, não apenas como aparecem no Excel.

tipos = {
    "Transaction_ID": "Identificador",
    "Order_ID": "Identificador",
    "Customer_ID": "Identificador",
    "Customer_Name": "Identificador",
    "Customer_Age": "Quantitativa",
    "Customer_Gender": "Categórica",
    "Customer_Segment": "Categórica",
    "Order_Date": "Data",
    "Order_Time": "Hora",
    "Sales_Channel": "Categórica",
    "Store_ID": "Identificador",
    "Store_Name": "Categórica",
    "Country": "Categórica",
    "Region": "Categórica",
    "City": "Categórica",
    "Product_ID": "Identificador",
    "Product_Name": "Categórica",
    "Product_Category": "Categórica",
    "Product_Subcategory": "Categórica",
    "Quantity": "Quantitativa",
    "Unit_Price": "Quantitativa",
    "Discount_Percentage": "Quantitativa",
    "Sales_Amount": "Quantitativa",
    "Cost_Amount": "Quantitativa",
    "Profit": "Quantitativa",
    "Payment_Method": "Categórica",
    "Order_Status": "Categórica",
    "Shipping_Method": "Categórica",
    "Delivery_Days": "Quantitativa",
    "Return_Flag": "Categórica",
    "Return_Reason": "Categórica",
    "Sales_Representative": "Categórica",
    "Promotion_Code": "Categórica",
    "Customer_Rating": "Quantitativa",
    "Inventory_Level": "Quantitativa",
    "Order_Year": "Ano"
}

# Dicionário com o tipo de cada coluna.
dicionario = pd.DataFrame({
    "Variável": df.columns,
    "Tipo": [tipos.get(c, "Verificar") for c in df.columns]
})

print("\nCLASSIFICAÇÃO DAS VARIÁVEIS:")
display(dicionario)


# ============================================================
# ETAPA 6 — PREPARAÇÃO PARA A ANÁLISE
# ============================================================

# Células que contêm somente espaços serão consideradas
# ausentes. Não substituímos os valores ausentes.

df = df.replace(r"^\s*$", np.nan, regex=True)

numericas = [
    c for c in df.columns
    if tipos.get(c) in ["Quantitativa", "Ano"]
]

for coluna in numericas:
    df[coluna] = pd.to_numeric(
        df[coluna], errors="coerce"
    )

if "Order_Date" in df.columns:
    df["Order_Date"] = pd.to_datetime(
        df["Order_Date"], errors="coerce"
    )

print("\nTipos de dados reconhecidos:")
display(df.dtypes.to_frame("Tipo no Python"))


# ============================================================
# ETAPA 7 — IDENTIFICAR MISSING VALUES
# ============================================================

print("\nANÁLISE DE VALORES AUSENTES")

missing = pd.DataFrame({
    "Variável": df.columns,
    "Ausentes": df.isna().sum().values,
    "Percentual (%)": (
        df.isna().mean().values * 100
    ).round(2)
})

display(missing)

print("\nTotal de células ausentes:",
      int(df.isna().sum().sum()))

# Gráfico dos valores ausentes

ausentes_grafico = missing[
    missing["Ausentes"] > 0
]

if not ausentes_grafico.empty:
    plt.figure(figsize=(12, 6))

    plt.barh(
        ausentes_grafico["Variável"],
        ausentes_grafico["Percentual (%)"]
    )

    plt.title("Percentual de valores ausentes")
    plt.xlabel("Percentual (%)")
    plt.tight_layout()
    plt.show()
else:
    print("Nenhuma ausência identificada.")


# ============================================================
# ETAPA 8 — FUNÇÃO PARA ANALISAR CADA VARIÁVEL
# ============================================================

def analisar_variavel(coluna, graficos=True):

    if coluna not in df.columns:
        print("Variável não encontrada:", coluna)
        return

    serie = df[coluna]
    tipo = tipos.get(coluna, "Verificar")

    print("\n" + "=" * 65)
    print("VARIÁVEL:", coluna)
    print("TIPO:", tipo)
    print("=" * 65)

    # Informações gerais

    print("Total de observações:", len(serie))
    print("Valores válidos:", serie.notna().sum())
    print("Valores ausentes:", serie.isna().sum())
    print("Percentual ausente:",
          round(serie.isna().mean() * 100, 2), "%")
    print("Valores distintos:", serie.nunique())

    # --------------------------------------------------------
    # ANÁLISE DE VARIÁVEIS QUANTITATIVAS
    # --------------------------------------------------------

    if tipo in ["Quantitativa", "Ano"]:

        dados = pd.to_numeric(
            serie, errors="coerce"
        ).dropna()

        if dados.empty:
            print("Não há valores numéricos válidos.")
            return

        print("\nESTATÍSTICA DESCRITIVA")

        estatisticas = pd.DataFrame({
            "Medida": [
                "Contagem",
                "Média",
                "Mediana",
                "Desvio padrão",
                "Variância",
                "Mínimo",
                "Primeiro quartil (Q1)",
                "Terceiro quartil (Q3)",
                "Máximo",
                "Assimetria",
                "Curtose",
                "Quantidade de zeros"
            ],
            "Resultado": [
                dados.count(),
                dados.mean(),
                dados.median(),
                dados.std(),
                dados.var(),
                dados.min(),
                dados.quantile(0.25),
                dados.quantile(0.75),
                dados.max(),
                dados.skew(),
                dados.kurt(),
                (dados == 0).sum()
            ]
        })

        display(estatisticas)

        # ----------------------------------------------------
        # DETECÇÃO DE OUTLIERS — MÉTODO IQR
        # ----------------------------------------------------

        q1 = dados.quantile(0.25)
        q3 = dados.quantile(0.75)

        iqr = q3 - q1

        limite_inferior = q1 - 1.5 * iqr
        limite_superior = q3 + 1.5 * iqr

        outliers = dados[
            (dados < limite_inferior) |
            (dados > limite_superior)
        ]

        print("\nANÁLISE DE OUTLIERS")

        print("IQR:", round(iqr, 3))
        print("Limite inferior:",
              round(limite_inferior, 3))
        print("Limite superior:",
              round(limite_superior, 3))

        print("Possíveis outliers:", len(outliers))

        print("Percentual de outliers:",
              round(
                  len(outliers) / len(dados) * 100, 2
              ), "%")

        # Não removemos outliers automaticamente.

        if graficos:

            # HISTOGRAMA

            plt.figure(figsize=(10, 4))
            plt.hist(
                dados,
                bins=30,
                edgecolor="black"
            )

            plt.axvline(
                dados.mean(),
                color="red",
                linestyle="--",
                label="Média"
            )

            plt.axvline(
                dados.median(),
                color="green",
                linestyle="--",
                label="Mediana"
            )

            plt.title("Histograma — " + coluna)
            plt.xlabel(coluna)
            plt.ylabel("Frequência")
            plt.legend()
            plt.tight_layout()
            plt.show()

            # BOXPLOT

            plt.figure(figsize=(10, 3))

            plt.boxplot(
                dados,
                vert=False
            )

            plt.title("Boxplot — " + coluna)
            plt.xlabel(coluna)
            plt.tight_layout()
            plt.show()

        return

    # --------------------------------------------------------
    # ANÁLISE DE VARIÁVEIS DE DATA
    # --------------------------------------------------------

    if tipo == "Data":

        datas = pd.to_datetime(
            serie, errors="coerce"
        ).dropna()

        if datas.empty:
            print("Não há datas válidas.")
            return

        print("\nPrimeira data:", datas.min())
        print("Última data:", datas.max())

        frequencia_mensal = (
            datas.dt.to_period("M")
            .value_counts()
            .sort_index()
        )

        display(frequencia_mensal.to_frame("Quantidade"))

        if graficos:
            frequencia_mensal.plot(
                kind="line",
                figsize=(12, 4),
                marker="o"
            )

            plt.title("Registros por mês")
            plt.xlabel("Mês")
            plt.ylabel("Frequência")
            plt.tight_layout()
            plt.show()

        return

    # --------------------------------------------------------
    # ANÁLISE DE HORÁRIOS
    # --------------------------------------------------------

    if tipo == "Hora":

        horarios = pd.to_datetime(
            serie.dropna().astype(str),
            format="%H:%M:%S",
            errors="coerce"
        )

        horas = horarios.dt.hour

        frequencias = (
            horas.value_counts().sort_index()
        )

        display(frequencias.to_frame("Frequência"))

        print("Horários não reconhecidos:",
              int(horas.isna().sum()))

        if graficos and not frequencias.empty:
            frequencias.plot(
                kind="bar",
                figsize=(10, 4)
            )

            plt.title("Distribuição por hora")
            plt.xlabel("Hora")
            plt.ylabel("Frequência")
            plt.tight_layout()
            plt.show()

        return

    # --------------------------------------------------------
    # ANÁLISE DE IDENTIFICADORES E CATEGORIAS
    # --------------------------------------------------------

    print("\nDISTRIBUIÇÃO DE FREQUÊNCIAS")

    frequencia = serie.value_counts(
        dropna=False
    )

    percentual = (
        frequencia / len(serie) * 100
    ).round(2)

    tabela = pd.DataFrame({
        "Frequência absoluta": frequencia,
        "Frequência relativa (%)": percentual
    })

    display(tabela.head(20))

    if tipo == "Identificador":
        print("\nIdentificadores repetidos:",
              serie.dropna().duplicated().sum())

        print(
            "Repetições podem ser legítimas, "
            "dependendo da unidade de análise."
        )

    elif graficos and not tabela.empty:

        plt.figure(figsize=(11, 5))

        frequencia.head(15).iloc[::-1].plot(
            kind="barh"
        )

        plt.title("Frequências — " + coluna)
        plt.xlabel("Quantidade")
        plt.tight_layout()
        plt.show()


# ============================================================
# ETAPA 9 — ANALISAR TODAS AS VARIÁVEIS
# ============================================================

# True: mostra gráficos.
# False: mostra resultados sem gráficos.

MOSTRAR_GRAFICOS = True

for numero, coluna in enumerate(df.columns, 1):

    print("\n")
    print("ANÁLISE", numero, "DE", len(df.columns))

    analisar_variavel(
        coluna,
        graficos=MOSTRAR_GRAFICOS
    )


# ============================================================
# ETAPA 10 — ANALISAR UMA VARIÁVEL ESPECÍFICA
# ============================================================

# Você pode trocar "Profit" por qualquer coluna.
# Exemplos:
# "Customer_Age"
# "Sales_Amount"
# "Discount_Percentage"
# "Quantity"

VARIAVEL_ESCOLHIDA = "Profit"

analisar_variavel(
    VARIAVEL_ESCOLHIDA,
    graficos=True
)


# ============================================================
# ETAPA 11 — VERIFICAR CONSISTÊNCIA DOS DADOS
# ============================================================

alertas = []

def registrar_alerta(nome, condicao):

    alertas.append({
        "Verificação": nome,
        "Ocorrências": int(
            condicao.fillna(False).sum()
        )
    })


# Verificação de valores negativos

colunas_verificar = [
    "Customer_Age",
    "Quantity",
    "Unit_Price",
    "Sales_Amount",
    "Cost_Amount",
    "Delivery_Days",
    "Inventory_Level"
]

for coluna in colunas_verificar:

    if coluna in df.columns:

        registrar_alerta(
            coluna + " negativo",
            df[coluna] < 0
        )


# Idades potencialmente inconsistentes

if "Customer_Age" in df.columns:

    registrar_alerta(
        "Idade fora de 0 a 120",
        (df["Customer_Age"] < 0) |
        (df["Customer_Age"] > 120)
    )


# Descontos fora da faixa esperada.
# Pressupõe desconto registrado entre 0 e 100.

if "Discount_Percentage" in df.columns:

    registrar_alerta(
        "Desconto fora de 0 a 100",
        (df["Discount_Percentage"] < 0) |
        (df["Discount_Percentage"] > 100)
    )


# Verificar igualdade contábil:
# Lucro = Vendas - Custos
# A regra deve ser confirmada no negócio.

colunas_lucro = [
    "Profit", "Sales_Amount", "Cost_Amount"
]

if all(c in df.columns for c in colunas_lucro):

    diferenca = (
        df["Sales_Amount"] -
        df["Cost_Amount"] -
        df["Profit"]
    ).abs()

    registrar_alerta(
        "Diferença na relação vendas-custo-lucro",
        diferenca > 0.02
    )


# Consistência entre data e ano do pedido

if {"Order_Date", "Order_Year"}.issubset(df.columns):

    ano_data = df["Order_Date"].dt.year

    divergencia = (
        ano_data.ne(df["Order_Year"]) &
        ano_data.notna() &
        df["Order_Year"].notna()
    )

    registrar_alerta(
        "Ano divergente da data do pedido",
        divergencia
    )


alertas_df = pd.DataFrame(alertas)

print("\nRELATÓRIO DE CONSISTÊNCIA")
display(alertas_df)


# ============================================================
# ETAPA 12 — RESUMO DA QUALIDADE DE CADA VARIÁVEL
# ============================================================

resumo = []

for coluna in df.columns:

    serie = df[coluna]

    item = {
        "Variável": coluna,
        "Tipo": tipos.get(coluna, "Verificar"),
        "Observações": len(serie),
        "Ausentes": int(serie.isna().sum()),
        "Ausentes (%)": round(
            serie.isna().mean() * 100, 2
        ),
        "Valores distintos": serie.nunique()
    }

    if coluna in numericas:

        dados = serie.dropna()

        if not dados.empty:

            q1 = dados.quantile(0.25)
            q3 = dados.quantile(0.75)

            iqr = q3 - q1

            li = q1 - 1.5 * iqr
            ls = q3 + 1.5 * iqr

            item["Outliers IQR"] = int(
                ((dados < li) | (dados > ls)).sum()
            )

    resumo.append(item)

resumo_df = pd.DataFrame(resumo)

print("\nRESUMO COMPLETO DAS VARIÁVEIS")
display(resumo_df)

print(
    "\nLinhas completamente duplicadas:",
    int(df.duplicated().sum())
)


# ============================================================
# ETAPA 13 — ANÁLISE DE DEVOLUÇÕES
# ============================================================

if {"Return_Flag", "Return_Reason"}.issubset(df.columns):

    print("\nDEVOLUÇÕES E MOTIVOS AUSENTES")

    tabela_devolucao = pd.crosstab(
        df["Return_Flag"].fillna("Ausente"),
        df["Return_Reason"].isna().map({
            True: "Motivo ausente",
            False: "Motivo informado"
        }),
        margins=True
    )

    display(tabela_devolucao)


# ============================================================
# ETAPA 14 — EXPORTAR RELATÓRIO PARA EXCEL
# ============================================================

nome_saida = "Relatorio_Analise_Exploratoria.xlsx"

with pd.ExcelWriter(
    nome_saida,
    engine="openpyxl"
) as writer:

    dicionario.to_excel(
        writer,
        sheet_name="Dicionario",
        index=False
    )

    missing.to_excel(
        writer,
        sheet_name="Missing_Values",
        index=False
    )

    resumo_df.to_excel(
        writer,
        sheet_name="Resumo_Variaveis",
        index=False
    )

    alertas_df.to_excel(
        writer,
        sheet_name="Consistencia",
        index=False
    )

    if {"Return_Flag", "Return_Reason"}.issubset(
        df.columns
    ):
        tabela_devolucao.to_excel(
            writer,
            sheet_name="Devolucoes"
        )

print("\nRELATÓRIO GERADO COM SUCESSO!")
print("Arquivo:", nome_saida)

files.download(nome_saida)

print("\nANÁLISE FINALIZADA!")
print("Os dados originais não foram alterados.")
```

```python

# ============================================================
# MACHINE LEARNING - PREVISÃO DO VALOR DAS VENDAS
# Dataset: Sales_transactions_2022_2025
#
# Algoritmos:
# 1. Regressão Linear
# 2. Lasso - Regularização L1
# 3. Ridge - Regularização L2
# 4. Elastic Net - Regularização L1 + L2
#
# Método:
# Pipeline + GridSearchCV + K-Fold (5 divisões)
#
# IMPORTANTE:
# O arquivo Excel original não será alterado.
# ============================================================


# ============================================================
# ETAPA 1 - IMPORTAR AS BIBLIOTECAS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import drive, files
from IPython.display import display

from sklearn.model_selection import (
    train_test_split,
    KFold,
    GridSearchCV
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.linear_model import (
    LinearRegression,
    Lasso,
    Ridge,
    ElasticNet
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Bibliotecas carregadas com sucesso!")


# ============================================================
# ETAPA 2 - CONECTAR AO GOOGLE DRIVE
# ============================================================

# O Google solicitará autorização de acesso.

drive.mount("/content/drive")


# ============================================================
# ETAPA 3 - CAMINHO DO ARQUIVO EXCEL
#
# ATENÇÃO! ALTERE SOMENTE A LINHA ABAIXO.
#
# Copie o caminho do arquivo que está no seu Drive
# e cole entre as aspas.
#
# Exemplo, se o arquivo estiver na pasta Dados:
#
# "/content/drive/MyDrive/Dados/arquivo.xlsx"
# ============================================================

CAMINHO_ARQUIVO = "/content/drive/MyDrive/Sales_transactions_2022_2025(1).xlsx"

# ============================================================
# FIM DA PARTE QUE VOCÊ PRECISA ALTERAR
# ============================================================


# ============================================================
# ETAPA 4 - CARREGAR O DATASET
# ============================================================

# Lê o arquivo original.
df_original = pd.read_excel(CAMINHO_ARQUIVO)

# Cria uma cópia independente para a análise.
df = df_original.copy(deep=True)

print("\nDATASET CARREGADO!")

print("Número de observações:", df.shape[0])

print("Número de variáveis:", df.shape[1])

print("\nPrimeiras cinco observações:")

display(df.head())


# ============================================================
# ETAPA 5 - DEFINIR A VARIÁVEL DEPENDENTE
# ============================================================

# Variável que queremos prever:
# Sales_Amount = valor da venda.

variavel_alvo = "Sales_Amount"


# ============================================================
# ETAPA 6 - DEFINIR OS OITO PREDITORES
# ============================================================

# Variáveis quantitativas:

preditores_numericos = [
    "Customer_Age",
    "Unit_Price",
    "Quantity",
    "Discount_Percentage",
    "Customer_Rating"
]

# Variáveis qualitativas nominais:

preditores_categoricos = [
    "Customer_Segment",
    "Sales_Channel",
    "Product_Category"
]

# Relação completa de preditores:

preditores = (
    preditores_numericos +
    preditores_categoricos
)

print("\nVARIÁVEL DEPENDENTE:")
print(variavel_alvo)

print("\nPREDITORES:")
for nome in preditores:
    print("-", nome)


# ============================================================
# ETAPA 7 - VERIFICAR A EXISTÊNCIA DAS COLUNAS
# ============================================================

colunas_necessarias = (
    preditores + [variavel_alvo]
)

colunas_faltantes = [
    coluna
    for coluna in colunas_necessarias
    if coluna not in df.columns
]

if colunas_faltantes:
    raise ValueError(
        "Colunas não encontradas: "
        + str(colunas_faltantes)
    )


# ============================================================
# ETAPA 8 - PREPARAR UMA CÓPIA DOS DADOS
# ============================================================

# Selecionamos somente as nove variáveis necessárias.
# O arquivo original permanece intacto.

dados = df[colunas_necessarias].copy()

# Converter colunas numéricas para formato numérico.
# Valores não interpretáveis são marcados como ausentes.

for coluna in (
    preditores_numericos + [variavel_alvo]
):
    dados[coluna] = pd.to_numeric(
        dados[coluna],
        errors="coerce"
    )

# Nas variáveis categóricas, células vazias
# ou preenchidas apenas com espaços viram ausentes.

for coluna in preditores_categoricos:
    dados[coluna] = (
        dados[coluna]
        .replace(r"^\s*$", np.nan, regex=True)
        .astype("object")
    )

# A regressão supervisionada precisa conhecer
# o valor real da variável dependente.
#
# Portanto, removemos da cópia somente registros
# que não possuem Sales_Amount válido.

total_antes = len(dados)

dados = dados.dropna(
    subset=[variavel_alvo]
).copy()

total_depois = len(dados)

print("\nREGISTROS EXCLUÍDOS DA CÓPIA POR ALVO AUSENTE:")
print(total_antes - total_depois)

# A planilha original não sofreu alterações.


# ============================================================
# ETAPA 9 - ANÁLISE DESCRITIVA INICIAL
# ============================================================

print("\nESTATÍSTICA DESCRITIVA:")

display(
    dados[
        preditores_numericos + [variavel_alvo]
    ].describe().T
)

print("\nVALORES AUSENTES:")

tabela_ausentes = pd.DataFrame({
    "Ausentes": dados.isna().sum(),
    "Percentual (%)": (
        dados.isna().mean() * 100
    ).round(2)
})

display(tabela_ausentes)


# ============================================================
# ETAPA 10 - SEPARAR X E y
# ============================================================

# X = variáveis explicativas.
# y = variável que queremos prever.

X = dados[preditores].copy()

y = dados[variavel_alvo].copy()

print("\nDimensões de X:", X.shape)
print("Dimensões de y:", y.shape)


# ============================================================
# ETAPA 11 - DIVIDIR TREINAMENTO E TESTE
# ============================================================

# 80% para desenvolver e selecionar o modelo.
# 20% reservados para o teste final.
#
# random_state=42 permite reproduzir a divisão.

X_treino, X_teste, y_treino, y_teste = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42
    )
)

print("\nDIVISÃO DOS DADOS:")

print("Treinamento:", len(X_treino))
print("Teste:", len(X_teste))

# IMPORTANTE:
# O teste não será usado para ajustar imputações,
# escalonamento ou escolher hiperparâmetros.


# ============================================================
# ETAPA 12 - PIPELINE DAS VARIÁVEIS NUMÉRICAS
# ============================================================

# Passo 1: substituir valores ausentes pela mediana.
# Passo 2: padronizar as variáveis pelo StandardScaler.
#
# A mediana e os parâmetros de padronização serão
# calculados somente nos dados de treinamento.

pipeline_numerico = Pipeline(
    steps=[
        (
            "imputacao",
            SimpleImputer(strategy="median")
        ),
        (
            "padronizacao",
            StandardScaler()
        )
    ]
)


# ============================================================
# ETAPA 13 - PIPELINE DAS VARIÁVEIS CATEGÓRICAS
# ============================================================

# Passo 1: imputar valores ausentes pela moda.
# Passo 2: aplicar One-Hot Encoding.
#
# Uma categoria nominal não possui necessariamente
# ordem numérica. O One-Hot evita impor essa ordem.
#
# handle_unknown="ignore" evita erro quando uma
# categoria nova aparece no conjunto de teste.

pipeline_categorico = Pipeline(
    steps=[
        (
            "imputacao",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)


# ============================================================
# ETAPA 14 - UNIR OS PRÉ-PROCESSAMENTOS
# ============================================================

preprocessador = ColumnTransformer(
    transformers=[
        (
            "numericas",
            pipeline_numerico,
            preditores_numericos
        ),
        (
            "categoricas",
            pipeline_categorico,
            preditores_categoricos
        )
    ]
)

print("\nPré-processador configurado!")


# ============================================================
# ETAPA 15 - DEFINIR O PIPELINE COMPLETO
# ============================================================

# O pipeline executará:
#
# 1. Imputação de missing values
# 2. Padronização das variáveis numéricas
# 3. One-Hot Encoding das categóricas
# 4. Ajuste do algoritmo de regressão

pipeline = Pipeline(
    steps=[
        (
            "preprocessamento",
            preprocessador
        ),
        (
            "modelo",
            LinearRegression()
        )
    ]
)


# ============================================================
# ETAPA 16 - DEFINIR OS QUATRO ALGORITMOS
# ============================================================

# O GridSearchCV testará:
#
# REGRESSÃO LINEAR:
# Sem penalização.
#
# LASSO:
# Regularização L1.
#
# RIDGE:
# Regularização L2.
#
# ELASTIC NET:
# Combinação das penalizações L1 e L2.
#
# alpha regula a intensidade da penalização.
# l1_ratio regula a composição L1/L2.

grade_parametros = [
    {
        "modelo": [
            LinearRegression()
        ]
    },

    {
        "modelo": [
            Lasso(
                max_iter=50000,
                random_state=42
            )
        ],
        "modelo__alpha": [
            0.001,
            0.01,
            0.1,
            1.0,
            10.0,
            100.0
        ]
    },

    {
        "modelo": [
            Ridge()
        ],
        "modelo__alpha": [
            0.001,
            0.01,
            0.1,
            1.0,
            10.0,
            100.0
        ]
    },

    {
        "modelo": [
            ElasticNet(
                max_iter=50000,
                random_state=42
            )
        ],
        "modelo__alpha": [
            0.001,
            0.01,
            0.1,
            1.0,
            10.0,
            100.0
        ],
        "modelo__l1_ratio": [
            0.2,
            0.5,
            0.8
        ]
    }
]


# ============================================================
# ETAPA 17 - CONFIGURAR VALIDAÇÃO CRUZADA K-FOLD
# ============================================================

# K = 5 significa cinco divisões.
#
# Em cada rodada:
# 4 partes são usadas para treinamento.
# 1 parte é usada para validação.
#
# Depois, ocorre uma troca de papéis entre as partes.

validacao = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ============================================================
# ETAPA 18 - COMPARAR OS ALGORITMOS
# ============================================================

# GridSearchCV:
# Compara os modelos e hiperparâmetros.
#
# scoring="neg_root_mean_squared_error"
# significa que queremos minimizar o RMSE.
#
# O sinal negativo é uma convenção do scikit-learn,
# porque sua rotina procura maximizar o score.
#
# refit=True ajusta novamente o melhor pipeline
# usando todos os dados de desenvolvimento.

busca = GridSearchCV(
    estimator=pipeline,
    param_grid=grade_parametros,
    scoring="neg_root_mean_squared_error",
    cv=validacao,
    refit=True,
    n_jobs=-1,
    verbose=1,
    error_score="raise"
)

print("\nINICIANDO VALIDAÇÃO CRUZADA...")
print("Aguarde o processamento dos modelos.")

busca.fit(
    X_treino,
    y_treino
)

print("\nVALIDAÇÃO CRUZADA CONCLUÍDA!")


# ============================================================
# ETAPA 19 - RESULTADOS DE TODOS OS MODELOS
# ============================================================

resultados_cv = pd.DataFrame(
    busca.cv_results_
)

# Identificar o nome de cada algoritmo.

def nome_algoritmo(modelo):

    return type(modelo).__name__


resultados_cv["Algoritmo"] = (
    resultados_cv["param_modelo"]
    .apply(nome_algoritmo)
)

# Converter RMSE negativo para valor positivo.

resultados_cv["RMSE_medio"] = (
    -resultados_cv["mean_test_score"]
)

resultados_cv["Desvio_RMSE"] = (
    resultados_cv["std_test_score"]
)

# Classificar do menor RMSE para o maior.

resultados_ordenados = (
    resultados_cv
    .sort_values("RMSE_medio")
)

# Guardar a melhor configuração de cada algoritmo.

comparacao = (
    resultados_ordenados
    .drop_duplicates(subset="Algoritmo")
    [[
        "Algoritmo",
        "RMSE_medio",
        "Desvio_RMSE",
        "params"
    ]]
    .reset_index(drop=True)
)

print("\nCOMPARAÇÃO DOS QUATRO ALGORITMOS:")

display(comparacao)


# ============================================================
# ETAPA 20 - IDENTIFICAR O MELHOR ALGORITMO
# ============================================================

melhor_pipeline = busca.best_estimator_

melhor_algoritmo = nome_algoritmo(
    melhor_pipeline.named_steps["modelo"]
)

print("\n====================================")
print("MELHOR ALGORITMO:", melhor_algoritmo)

print(
    "RMSE médio da validação:",
    round(-busca.best_score_, 4)
)

print("\nMelhores parâmetros:")

print(busca.best_params_)

print("====================================")


# ============================================================
# ETAPA 21 - REALIZAR O TESTE FINAL
# ============================================================

# Só agora utilizamos os 20% reservados.
#
# Nenhum parâmetro do modelo será ajustado
# utilizando essas observações.

y_previsto = melhor_pipeline.predict(
    X_teste
)

# MAE = erro absoluto médio.
mae = mean_absolute_error(
    y_teste,
    y_previsto
)

# RMSE = raiz do erro quadrático médio.
rmse = np.sqrt(
    mean_squared_error(
        y_teste,
        y_previsto
    )
)

# R² = coeficiente de determinação.
r2 = r2_score(
    y_teste,
    y_previsto
)

# WAPE = soma dos erros absolutos dividida
# pela soma dos valores absolutos observados.

denominador = np.abs(y_teste).sum()

wape = (
    np.abs(y_teste - y_previsto).sum()
    / denominador * 100
    if denominador > 0
    else np.nan
)

print("\nRESULTADOS DO TESTE FINAL:")

print("MAE:", round(mae, 4))
print("RMSE:", round(rmse, 4))
print("R²:", round(r2, 4))
print("WAPE (%):", round(wape, 2))


# ============================================================
# ETAPA 22 - GRÁFICO COMPARATIVO DOS ALGORITMOS
# ============================================================

plt.figure(figsize=(10, 5))

plt.bar(
    comparacao["Algoritmo"],
    comparacao["RMSE_medio"]
)

plt.ylabel("RMSE médio na validação")
plt.xlabel("Algoritmo")
plt.title("Comparação dos modelos de regressão")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ============================================================
# ETAPA 23 - COMPARAR VENDAS REAIS E PREVISTAS
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_teste,
    y_previsto,
    alpha=0.45
)

menor = min(
    y_teste.min(),
    y_previsto.min()
)

maior = max(
    y_teste.max(),
    y_previsto.max()
)

# Linha de previsão perfeita.
plt.plot(
    [menor, maior],
    [menor, maior],
    "r--",
    label="Previsão perfeita"
)

plt.xlabel("Valor real da venda")
plt.ylabel("Valor previsto da venda")

plt.title(
    "Valores reais versus valores previstos"
)

plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# ETAPA 24 - ANÁLISE DOS RESÍDUOS
# ============================================================

# Resíduo = valor real - valor previsto.
#
# Se o valor é positivo:
# o modelo subestimou a venda.
#
# Se o valor é negativo:
# o modelo superestimou a venda.

residuos = y_teste - y_previsto

print("\nESTATÍSTICA DESCRITIVA DOS RESÍDUOS:")

display(
    pd.Series(residuos).describe()
)

# Histograma dos resíduos

plt.figure(figsize=(10, 5))

plt.hist(
    residuos,
    bins=40,
    edgecolor="black"
)

plt.axvline(
    0,
    color="red",
    linestyle="--"
)

plt.title("Distribuição dos resíduos")
plt.xlabel("Resíduo")
plt.ylabel("Frequência")

plt.tight_layout()
plt.show()


# Resíduos versus valores previstos

plt.figure(figsize=(10, 5))

plt.scatter(
    y_previsto,
    residuos,
    alpha=0.45
)

plt.axhline(
    0,
    color="red",
    linestyle="--"
)

plt.xlabel("Valor previsto")
plt.ylabel("Resíduo")
plt.title("Resíduos versus previsões")

plt.tight_layout()
plt.show()


# ============================================================
# ETAPA 25 - ANALISAR OS COEFICIENTES
# ============================================================

# Como os modelos são lineares, é possível verificar
# os coeficientes associados a cada variável após
# o pré-processamento.

nomes_variaveis = (
    melhor_pipeline
    .named_steps["preprocessamento"]
    .get_feature_names_out()
)

coeficientes = (
    melhor_pipeline
    .named_steps["modelo"]
    .coef_
)

tabela_coeficientes = pd.DataFrame({
    "Variável transformada": nomes_variaveis,
    "Coeficiente": coeficientes
})

tabela_coeficientes["Valor_absoluto"] = (
    tabela_coeficientes["Coeficiente"].abs()
)

tabela_coeficientes = (
    tabela_coeficientes
    .sort_values(
        "Valor_absoluto",
        ascending=False
    )
)

print("\nCOEFICIENTES DO MELHOR MODELO:")

display(
    tabela_coeficientes.drop(
        columns="Valor_absoluto"
    )
)

# Atenção:
# Os coeficientes numéricos correspondem às
# variáveis padronizadas.
#
# Coeficientes não demonstram causalidade.


# ============================================================
# ETAPA 26 - VERIFICAR A FÓRMULA COMERCIAL
# ============================================================

# Em muitos sistemas:
#
# Valor da venda =
# Preço unitário * Quantidade *
# (1 - Desconto percentual / 100)
#
# Vamos verificar se essa fórmula se aproxima
# do Sales_Amount registrado.
#
# Este é um diagnóstico separado dos modelos.
# Não altera a seleção do algoritmo.

formula_venda = (
    X_teste["Unit_Price"]
    * X_teste["Quantity"]
    * (
        1 -
        X_teste["Discount_Percentage"] / 100
    )
)

mascara_formula = (
    formula_venda.notna() &
    np.isfinite(formula_venda)
)

if mascara_formula.any():

    real_formula = y_teste.loc[
        mascara_formula
    ]

    previsto_formula = formula_venda.loc[
        mascara_formula
    ]

    erro_formula = mean_absolute_error(
        real_formula,
        previsto_formula
    )

    print("\nVERIFICAÇÃO DA FÓRMULA COMERCIAL:")

    print(
        "MAE da fórmula:",
        round(erro_formula, 4)
    )

    print(
        "Registros verificados:",
        int(mascara_formula.sum())
    )

    print(
        "Esta verificação não substitui "
        "a avaliação do modelo escolhido."
    )

else:
    print(
        "\nNão há observações suficientes "
        "para verificar a fórmula comercial."
    )


# ============================================================
# ETAPA 27 - GERAR TABELA DE PREVISÕES
# ============================================================

# Organizar os resultados do teste em uma tabela.

tabela_previsoes = X_teste.copy()

tabela_previsoes["Venda_Real"] = (
    y_teste
)

tabela_previsoes["Venda_Prevista"] = (
    y_previsto
)

tabela_previsoes["Residuo"] = (
    y_teste - y_previsto
)

tabela_previsoes["Erro_Absoluto"] = (
    np.abs(y_teste - y_previsto)
)

print("\nAMOSTRA DAS PREVISÕES:")

display(
    tabela_previsoes.head(20)
)


# ============================================================
# ETAPA 28 - EXPORTAR OS RESULTADOS PARA EXCEL
# ============================================================

nome_relatorio = (
    "Resultado_Machine_Learning_Vendas.xlsx"
)

metricas_finais = pd.DataFrame({
    "Indicador": [
        "Melhor algoritmo",
        "MAE",
        "RMSE",
        "R²",
        "WAPE (%)"
    ],
    "Resultado": [
        melhor_algoritmo,
        mae,
        rmse,
        r2,
        wape
    ]
})

with pd.ExcelWriter(
    nome_relatorio,
    engine="openpyxl"
) as writer:

    comparacao.to_excel(
        writer,
        sheet_name="Comparacao_Modelos",
        index=False
    )

    metricas_finais.to_excel(
        writer,
        sheet_name="Metricas_Teste",
        index=False
    )

    tabela_coeficientes.drop(
        columns="Valor_absoluto"
    ).to_excel(
        writer,
        sheet_name="Coeficientes",
        index=False
    )

    tabela_previsoes.to_excel(
        writer,
        sheet_name="Previsoes",
        index=False
    )

print("\nRELATÓRIO EXCEL GERADO!")

files.download(nome_relatorio)


# ============================================================
# ETAPA 29 - CONCLUSÃO
# ============================================================

print("\n====================================")
print("ANÁLISE FINALIZADA")
print("====================================")

print("Melhor algoritmo:", melhor_algoritmo)

print("RMSE no teste:", round(rmse, 4))

print("R² no teste:", round(r2, 4))

print("\nO DATASET ORIGINAL FOI PRESERVADO.")

print(
    "A seleção dos modelos foi realizada "
    "somente nos dados de desenvolvimento."
)
```

```python

# ============================================================
# MACHINE LEARNING - PREVISÃO DO VALOR DAS VENDAS
# Dataset: Sales_transactions_2022_2025
#
# Algoritmos:
# 1. Regressão Linear
# 2. Lasso - Regularização L1
# 3. Ridge - Regularização L2
# 4. Elastic Net - Regularização L1 + L2
#
# Método:
# Pipeline + GridSearchCV + K-Fold (5 divisões)
#
# IMPORTANTE:
# O arquivo Excel original não será alterado.
# ============================================================


# ============================================================
# ETAPA 1 - IMPORTAR AS BIBLIOTECAS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import drive, files
from IPython.display import display

from sklearn.model_selection import (
    train_test_split,
    KFold,
    GridSearchCV
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.linear_model import (
    LinearRegression,
    Lasso,
    Ridge,
    ElasticNet
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Bibliotecas carregadas com sucesso!")


# ============================================================
# ETAPA 2 - CONECTAR AO GOOGLE DRIVE
# ============================================================

# O Google solicitará autorização de acesso.

drive.mount("/content/drive")


# ============================================================
# ETAPA 3 - CAMINHO DO ARQUIVO EXCEL
#
# ATENÇÃO! ALTERE SOMENTE A LINHA ABAIXO.
#
# Copie o caminho do arquivo que está no seu Drive
# e cole entre as aspas.
#
# Exemplo, se o arquivo estiver na pasta Dados:
#
# "/content/drive/MyDrive/Dados/arquivo.xlsx"
# ============================================================

CAMINHO_ARQUIVO = "/content/drive/MyDrive/Banco de Dados 1/Sales_transactions_2022_2025.xlsx"

# ============================================================
# FIM DA PARTE QUE VOCÊ PRECISA ALTERAR
# ============================================================


# ============================================================
# ETAPA 4 - CARREGAR O DATASET
# ============================================================

# Lê o arquivo original.
df_original = pd.read_excel(CAMINHO_ARQUIVO)

# Cria uma cópia independente para a análise.
df = df_original.copy(deep=True)

print("\nDATASET CARREGADO!")

print("Número de observações:", df.shape[0])

print("Número de variáveis:", df.shape[1])

print("\nPrimeiras cinco observações:")

display(df.head())


# ============================================================
# ETAPA 5 - DEFINIR A VARIÁVEL DEPENDENTE
# ============================================================

# Variável que queremos prever:
# Sales_Amount = valor da venda.

variavel_alvo = "Sales_Amount"


# ============================================================
# ETAPA 6 - DEFINIR OS OITO PREDITORES
# ============================================================

# Variáveis quantitativas:

preditores_numericos = [
    "Customer_Age",
    "Unit_Price",
    "Quantity",
    "Discount_Percentage",
    "Customer_Rating"
]

# Variáveis qualitativas nominais:

preditores_categoricos = [
    "Customer_Segment",
    "Sales_Channel",
    "Product_Category"
]

# Relação completa de preditores:

preditores = (
    preditores_numericos +
    preditores_categoricos
)

print("\nVARIÁVEL DEPENDENTE:")
print(variavel_alvo)

print("\nPREDITORES:")
for nome in preditores:
    print("-", nome)


# ============================================================
# ETAPA 7 - VERIFICAR A EXISTÊNCIA DAS COLUNAS
# ============================================================

colunas_necessarias = (
    preditores + [variavel_alvo]
)

colunas_faltantes = [
    coluna
    for coluna in colunas_necessarias
    if coluna not in df.columns
]

if colunas_faltantes:
    raise ValueError(
        "Colunas não encontradas: "
        + str(colunas_faltantes)
    )


# ============================================================
# ETAPA 8 - PREPARAR UMA CÓPIA DOS DADOS
# ============================================================

# Selecionamos somente as nove variáveis necessárias.
# O arquivo original permanece intacto.

dados = df[colunas_necessarias].copy()

# Converter colunas numéricas para formato numérico.
# Valores não interpretáveis são marcados como ausentes.

for coluna in (
    preditores_numericos + [variavel_alvo]
):
    dados[coluna] = pd.to_numeric(
        dados[coluna],
        errors="coerce"
    )

# Nas variáveis categóricas, células vazias
# ou preenchidas apenas com espaços viram ausentes.

for coluna in preditores_categoricos:
    dados[coluna] = (
        dados[coluna]
        .replace(r"^\s*$", np.nan, regex=True)
        .astype("object")
    )

# A regressão supervisionada precisa conhecer
# o valor real da variável dependente.
#
# Portanto, removemos da cópia somente registros
# que não possuem Sales_Amount válido.

total_antes = len(dados)

dados = dados.dropna(
    subset=[variavel_alvo]
).copy()

total_depois = len(dados)

print("\nREGISTROS EXCLUÍDOS DA CÓPIA POR ALVO AUSENTE:")
print(total_antes - total_depois)

# A planilha original não sofreu alterações.


# ============================================================
# ETAPA 9 - ANÁLISE DESCRITIVA INICIAL
# ============================================================

print("\nESTATÍSTICA DESCRITIVA:")

display(
    dados[
        preditores_numericos + [variavel_alvo]
    ].describe().T
)

print("\nVALORES AUSENTES:")

tabela_ausentes = pd.DataFrame({
    "Ausentes": dados.isna().sum(),
    "Percentual (%)": (
        dados.isna().mean() * 100
    ).round(2)
})

display(tabela_ausentes)


# ============================================================
# ETAPA 10 - SEPARAR X E y
# ============================================================

# X = variáveis explicativas.
# y = variável que queremos prever.

X = dados[preditores].copy()

y = dados[variavel_alvo].copy()

print("\nDimensões de X:", X.shape)
print("Dimensões de y:", y.shape)


# ============================================================
# ETAPA 11 - DIVIDIR TREINAMENTO E TESTE
# ============================================================

# 80% para desenvolver e selecionar o modelo.
# 20% reservados para o teste final.
#
# random_state=42 permite reproduzir a divisão.

X_treino, X_teste, y_treino, y_teste = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42
    )
)

print("\nDIVISÃO DOS DADOS:")

print("Treinamento:", len(X_treino))
print("Teste:", len(X_teste))

# IMPORTANTE:
# O teste não será usado para ajustar imputações,
# escalonamento ou escolher hiperparâmetros.


# ============================================================
# ETAPA 12 - PIPELINE DAS VARIÁVEIS NUMÉRICAS
# ============================================================

# Passo 1: substituir valores ausentes pela mediana.
# Passo 2: padronizar as variáveis pelo StandardScaler.
#
# A mediana e os parâmetros de padronização serão
# calculados somente nos dados de treinamento.

pipeline_numerico = Pipeline(
    steps=[
        (
            "imputacao",
            SimpleImputer(strategy="median")
        ),
        (
            "padronizacao",
            StandardScaler()
        )
    ]
)


# ============================================================
# ETAPA 13 - PIPELINE DAS VARIÁVEIS CATEGÓRICAS
# ============================================================

# Passo 1: imputar valores ausentes pela moda.
# Passo 2: aplicar One-Hot Encoding.
#
# Uma categoria nominal não possui necessariamente
# ordem numérica. O One-Hot evita impor essa ordem.
#
# handle_unknown="ignore" evita erro quando uma
# categoria nova aparece no conjunto de teste.

pipeline_categorico = Pipeline(
    steps=[
        (
            "imputacao",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)


# ============================================================
# ETAPA 14 - UNIR OS PRÉ-PROCESSAMENTOS
# ============================================================

preprocessador = ColumnTransformer(
    transformers=[
        (
            "numericas",
            pipeline_numerico,
            preditores_numericos
        ),
        (
            "categoricas",
            pipeline_categorico,
            preditores_categoricos
        )
    ]
)

print("\nPré-processador configurado!")


# ============================================================
# ETAPA 15 - DEFINIR O PIPELINE COMPLETO
# ============================================================

# O pipeline executará:
#
# 1. Imputação de missing values
# 2. Padronização das variáveis numéricas
# 3. One-Hot Encoding das categóricas
# 4. Ajuste do algoritmo de regressão

pipeline = Pipeline(
    steps=[
        (
            "preprocessamento",
            preprocessador
        ),
        (
            "modelo",
            LinearRegression()
        )
    ]
)


# ============================================================
# ETAPA 16 - DEFINIR OS QUATRO ALGORITMOS
# ============================================================

# O GridSearchCV testará:
#
# REGRESSÃO LINEAR:
# Sem penalização.
#
# LASSO:
# Regularização L1.
#
# RIDGE:
# Regularização L2.
#
# ELASTIC NET:
# Combinação das penalizações L1 e L2.
#
# alpha regula a intensidade da penalização.
# l1_ratio regula a composição L1/L2.

grade_parametros = [
    {
        "modelo": [
            LinearRegression()
        ]
    },

    {
        "modelo": [
            Lasso(
                max_iter=50000,
                random_state=42
            )
        ],
        "modelo__alpha": [
            0.001,
            0.01,
            0.1,
            1.0,
            10.0,
            100.0
        ]
    },

    {
        "modelo": [
            Ridge()
        ],
        "modelo__alpha": [
            0.001,
            0.01,
            0.1,
            1.0,
            10.0,
            100.0
        ]
    },

    {
        "modelo": [
            ElasticNet(
                max_iter=50000,
                random_state=42
            )
        ],
        "modelo__alpha": [
            0.001,
            0.01,
            0.1,
            1.0,
            10.0,
            100.0
        ],
        "modelo__l1_ratio": [
            0.2,
            0.5,
            0.8
        ]
    }
]


# ============================================================
# ETAPA 17 - CONFIGURAR VALIDAÇÃO CRUZADA K-FOLD
# ============================================================

# K = 5 significa cinco divisões.
#
# Em cada rodada:
# 4 partes são usadas para treinamento.
# 1 parte é usada para validação.
#
# Depois, ocorre uma troca de papéis entre as partes.

validacao = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ============================================================
# ETAPA 18 - COMPARAR OS ALGORITMOS
# ============================================================

# GridSearchCV:
# Compara os modelos e hiperparâmetros.
#
# scoring="neg_root_mean_squared_error"
# significa que queremos minimizar o RMSE.
#
# O sinal negativo é uma convenção do scikit-learn,
# porque sua rotina procura maximizar o score.
#
# refit=True ajusta novamente o melhor pipeline
# usando todos os dados de desenvolvimento.

busca = GridSearchCV(
    estimator=pipeline,
    param_grid=grade_parametros,
    scoring="neg_root_mean_squared_error",
    cv=validacao,
    refit=True,
    n_jobs=-1,
    verbose=1,
    error_score="raise"
)

print("\nINICIANDO VALIDAÇÃO CRUZADA...")
print("Aguarde o processamento dos modelos.")

busca.fit(
    X_treino,
    y_treino
)

print("\nVALIDAÇÃO CRUZADA CONCLUÍDA!")


# ============================================================
# ETAPA 19 - RESULTADOS DE TODOS OS MODELOS
# ============================================================

resultados_cv = pd.DataFrame(
    busca.cv_results_
)

# Identificar o nome de cada algoritmo.

def nome_algoritmo(modelo):

    return type(modelo).__name__


resultados_cv["Algoritmo"] = (
    resultados_cv["param_modelo"]
    .apply(nome_algoritmo)
)

# Converter RMSE negativo para valor positivo.

resultados_cv["RMSE_medio"] = (
    -resultados_cv["mean_test_score"]
)

resultados_cv["Desvio_RMSE"] = (
    resultados_cv["std_test_score"]
)

# Classificar do menor RMSE para o maior.

resultados_ordenados = (
    resultados_cv
    .sort_values("RMSE_medio")
)

# Guardar a melhor configuração de cada algoritmo.

comparacao = (
    resultados_ordenados
    .drop_duplicates(subset="Algoritmo")
    [[
        "Algoritmo",
        "RMSE_medio",
        "Desvio_RMSE",
        "params"
    ]]
    .reset_index(drop=True)
)

print("\nCOMPARAÇÃO DOS QUATRO ALGORITMOS:")

display(comparacao)


# ============================================================
# ETAPA 20 - IDENTIFICAR O MELHOR ALGORITMO
# ============================================================

melhor_pipeline = busca.best_estimator_

melhor_algoritmo = nome_algoritmo(
    melhor_pipeline.named_steps["modelo"]
)

print("\n====================================")
print("MELHOR ALGORITMO:", melhor_algoritmo)

print(
    "RMSE médio da validação:",
    round(-busca.best_score_, 4)
)

print("\nMelhores parâmetros:")

print(busca.best_params_)

print("====================================")


# ============================================================
# ETAPA 21 - REALIZAR O TESTE FINAL
# ============================================================

# Só agora utilizamos os 20% reservados.
#
# Nenhum parâmetro do modelo será ajustado
# utilizando essas observações.

y_previsto = melhor_pipeline.predict(
    X_teste
)

# MAE = erro absoluto médio.
mae = mean_absolute_error(
    y_teste,
    y_previsto
)

# RMSE = raiz do erro quadrático médio.
rmse = np.sqrt(
    mean_squared_error(
        y_teste,
        y_previsto
    )
)

# R² = coeficiente de determinação.
r2 = r2_score(
    y_teste,
    y_previsto
)

# WAPE = soma dos erros absolutos dividida
# pela soma dos valores absolutos observados.

denominador = np.abs(y_teste).sum()

wape = (
    np.abs(y_teste - y_previsto).sum()
    / denominador * 100
    if denominador > 0
    else np.nan
)

print("\nRESULTADOS DO TESTE FINAL:")

print("MAE:", round(mae, 4))
print("RMSE:", round(rmse, 4))
print("R²:", round(r2, 4))
print("WAPE (%):", round(wape, 2))


# ============================================================
# ETAPA 22 - GRÁFICO COMPARATIVO DOS ALGORITMOS
# ============================================================

plt.figure(figsize=(10, 5))

plt.bar(
    comparacao["Algoritmo"],
    comparacao["RMSE_medio"]
)

plt.ylabel("RMSE médio na validação")
plt.xlabel("Algoritmo")
plt.title("Comparação dos modelos de regressão")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ============================================================
# ETAPA 23 - COMPARAR VENDAS REAIS E PREVISTAS
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_teste,
    y_previsto,
    alpha=0.45
)

menor = min(
    y_teste.min(),
    y_previsto.min()
)

maior = max(
    y_teste.max(),
    y_previsto.max()
)

# Linha de previsão perfeita.
plt.plot(
    [menor, maior],
    [menor, maior],
    "r--",
    label="Previsão perfeita"
)

plt.xlabel("Valor real da venda")
plt.ylabel("Valor previsto da venda")

plt.title(
    "Valores reais versus valores previstos"
)

plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# ETAPA 24 - ANÁLISE DOS RESÍDUOS
# ============================================================

# Resíduo = valor real - valor previsto.
#
# Se o valor é positivo:
# o modelo subestimou a venda.
#
# Se o valor é negativo:
# o modelo superestimou a venda.

residuos = y_teste - y_previsto

print("\nESTATÍSTICA DESCRITIVA DOS RESÍDUOS:")

display(
    pd.Series(residuos).describe()
)

# Histograma dos resíduos

plt.figure(figsize=(10, 5))

plt.hist(
    residuos,
    bins=40,
    edgecolor="black"
)

plt.axvline(
    0,
    color="red",
    linestyle="--"
)

plt.title("Distribuição dos resíduos")
plt.xlabel("Resíduo")
plt.ylabel("Frequência")

plt.tight_layout()
plt.show()


# Resíduos versus valores previstos

plt.figure(figsize=(10, 5))

plt.scatter(
    y_previsto,
    residuos,
    alpha=0.45
)

plt.axhline(
    0,
    color="red",
    linestyle="--"
)

plt.xlabel("Valor previsto")
plt.ylabel("Resíduo")
plt.title("Resíduos versus previsões")

plt.tight_layout()
plt.show()


# ============================================================
# ETAPA 25 - ANALISAR OS COEFICIENTES
# ============================================================

# Como os modelos são lineares, é possível verificar
# os coeficientes associados a cada variável após
# o pré-processamento.

nomes_variaveis = (
    melhor_pipeline
    .named_steps["preprocessamento"]
    .get_feature_names_out()
)

coeficientes = (
    melhor_pipeline
    .named_steps["modelo"]
    .coef_
)

tabela_coeficientes = pd.DataFrame({
    "Variável transformada": nomes_variaveis,
    "Coeficiente": coeficientes
})

tabela_coeficientes["Valor_absoluto"] = (
    tabela_coeficientes["Coeficiente"].abs()
)

tabela_coeficientes = (
    tabela_coeficientes
    .sort_values(
        "Valor_absoluto",
        ascending=False
    )
)

print("\nCOEFICIENTES DO MELHOR MODELO:")

display(
    tabela_coeficientes.drop(
        columns="Valor_absoluto"
    )
)

# Atenção:
# Os coeficientes numéricos correspondem às
# variáveis padronizadas.
#
# Coeficientes não demonstram causalidade.


# ============================================================
# ETAPA 26 - VERIFICAR A FÓRMULA COMERCIAL
# ============================================================

# Em muitos sistemas:
#
# Valor da venda =
# Preço unitário * Quantidade *
# (1 - Desconto percentual / 100)
#
# Vamos verificar se essa fórmula se aproxima
# do Sales_Amount registrado.
#
# Este é um diagnóstico separado dos modelos.
# Não altera a seleção do algoritmo.

formula_venda = (
    X_teste["Unit_Price"]
    * X_teste["Quantity"]
    * (
        1 -
        X_teste["Discount_Percentage"] / 100
    )
)

mascara_formula = (
    formula_venda.notna() &
    np.isfinite(formula_venda)
)

if mascara_formula.any():

    real_formula = y_teste.loc[
        mascara_formula
    ]

    previsto_formula = formula_venda.loc[
        mascara_formula
    ]

    erro_formula = mean_absolute_error(
        real_formula,
        previsto_formula
    )

    print("\nVERIFICAÇÃO DA FÓRMULA COMERCIAL:")

    print(
        "MAE da fórmula:",
        round(erro_formula, 4)
    )

    print(
        "Registros verificados:",
        int(mascara_formula.sum())
    )

    print(
        "Esta verificação não substitui "
        "a avaliação do modelo escolhido."
    )

else:
    print(
        "\nNão há observações suficientes "
        "para verificar a fórmula comercial."
    )


# ============================================================
# ETAPA 27 - GERAR TABELA DE PREVISÕES
# ============================================================

# Organizar os resultados do teste em uma tabela.

tabela_previsoes = X_teste.copy()

tabela_previsoes["Venda_Real"] = (
    y_teste
)

tabela_previsoes["Venda_Prevista"] = (
    y_previsto
)

tabela_previsoes["Residuo"] = (
    y_teste - y_previsto
)

tabela_previsoes["Erro_Absoluto"] = (
    np.abs(y_teste - y_previsto)
)

print("\nAMOSTRA DAS PREVISÕES:")

display(
    tabela_previsoes.head(20)
)


# ============================================================
# ETAPA 28 - EXPORTAR OS RESULTADOS PARA EXCEL
# ============================================================

nome_relatorio = (
    "Resultado_Machine_Learning_Vendas.xlsx"
)

metricas_finais = pd.DataFrame({
    "Indicador": [
        "Melhor algoritmo",
        "MAE",
        "RMSE",
        "R²",
        "WAPE (%)"
    ],
    "Resultado": [
        melhor_algoritmo,
        mae,
        rmse,
        r2,
        wape
    ]
})

with pd.ExcelWriter(
    nome_relatorio,
    engine="openpyxl"
) as writer:

    comparacao.to_excel(
        writer,
        sheet_name="Comparacao_Modelos",
        index=False
    )

    metricas_finais.to_excel(
        writer,
        sheet_name="Metricas_Teste",
        index=False
    )

    tabela_coeficientes.drop(
        columns="Valor_absoluto"
    ).to_excel(
        writer,
        sheet_name="Coeficientes",
        index=False
    )

    tabela_previsoes.to_excel(
        writer,
        sheet_name="Previsoes",
        index=False
    )

print("\nRELATÓRIO EXCEL GERADO!")

files.download(nome_relatorio)


# ============================================================
# ETAPA 29 - CONCLUSÃO
# ============================================================

print("\n====================================")
print("ANÁLISE FINALIZADA")
print("====================================")

print("Melhor algoritmo:", melhor_algoritmo)

print("RMSE no teste:", round(rmse, 4))

print("R² no teste:", round(r2, 4))

print("\nO DATASET ORIGINAL FOI PRESERVADO.")

print(
    "A seleção dos modelos foi realizada "
    "somente nos dados de desenvolvimento."
)
```
